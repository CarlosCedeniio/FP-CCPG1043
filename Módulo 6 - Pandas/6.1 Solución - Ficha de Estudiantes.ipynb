# Módulo 6.1 — Solución: Ficha de Estudiantes

Resolución paso a paso del **Caso Práctico** de la clase 6.1 (`Introducción a Pandas - Series y DataFrames`).

Construimos, desde cero, un DataFrame de estudiantes y lo exploramos con los métodos básicos de Pandas.

## Paso 1: Crear el diccionario y el DataFrame

Un diccionario con 6 estudiantes: `nombre`, `carrera`, `semestre`, `nota` y `asistencia`. Cada clave del diccionario se convierte en una columna.

```python
import pandas as pd

datos = {
    'nombre':     ['Ana', 'Bruno', 'Camila', 'Diego', 'Elena', 'Fabián'],
    'carrera':    ['Ingeniería', 'Medicina', 'Derecho', 'Administración', 'Arquitectura', 'Psicología'],
    'semestre':   [4, 2, 6, 3, 5, 1],
    'nota':       [9.03, 8.24, 7.51, 8.77, 6.88, 9.31],
    'asistencia': [95, 92, 78, 91, 65, 97]
}

mis_estudiantes = pd.DataFrame(datos)
mis_estudiantes
```

## Paso 2: Explorar con `.head()`, `.info()` y `.describe()`

Antes de hacer cualquier cosa con el DataFrame, conviene explorarlo primero.

```python
mis_estudiantes.head(3)
```

```python
mis_estudiantes.info()
```

```python
mis_estudiantes.describe()
```

`.info()` confirma que no hay nulos (las 6 columnas tienen 6 valores no nulos) y `.describe()` resume las columnas numéricas: promedio, mínimo, máximo, etc.

## Paso 3: Seleccionar solo `nombre` y `nota`

Usamos una lista de nombres de columna para quedarnos con un DataFrame más pequeño.

```python
nombre_y_nota = mis_estudiantes[['nombre', 'nota']]
nombre_y_nota
```

## Paso 4: Primera y última fila con `.loc` / `.iloc`

Como el índice es el numérico por defecto (0, 1, 2...), tanto `.loc` como `.iloc` funcionan aquí, pero por significado: `.iloc` es la forma correcta de decir "la primera" y "la última" fila **por posición**, sin importar qué etiqueta tenga el índice.

```python
mis_estudiantes.iloc[0]
```

```python
mis_estudiantes.iloc[-1]
```

## Paso 5: Agregar `nota_redondeada` con `.round()`

```python
mis_estudiantes['nota_redondeada'] = mis_estudiantes['nota'].round(1)
mis_estudiantes[['nombre', 'nota', 'nota_redondeada']]
```

## Paso 6: Agregar la columna booleana `beca`

`True` si `nota >= 8.5` **y** `asistencia >= 90`. Como son dos condiciones sobre columnas completas, se combinan con `&` (no con `and`), y cada condición va entre paréntesis.

```python
mis_estudiantes['beca'] = (mis_estudiantes['nota'] >= 8.5) & (mis_estudiantes['asistencia'] >= 90)
mis_estudiantes[['nombre', 'nota', 'asistencia', 'beca']]
```

Ana, Diego y Fabián cumplen ambas condiciones a la vez; Bruno tiene buena asistencia pero nota menor a 8.5; Camila y Elena no cumplen ninguna.

## Paso 7: Promedio de `nota` y de `asistencia`

```python
promedio_nota = mis_estudiantes['nota'].mean()
promedio_asistencia = mis_estudiantes['asistencia'].mean()

print(f'Promedio de nota: {promedio_nota:.2f}')
print(f'Promedio de asistencia: {promedio_asistencia:.2f}%')
```
